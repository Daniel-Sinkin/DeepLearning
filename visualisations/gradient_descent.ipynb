```python
import numpy as np
import plotly.graph_objects as go

def f(x: np.ndarray, y: np.ndarray) -> np.ndarray:
    return np.sin(x) * np.cos(y) + 0.1 * (x**2 + y**2)

def grad_f(x: np.ndarray, y: np.ndarray) -> tuple[np.ndarray, np.ndarray]:
    df_dx = np.cos(x) * np.cos(y) + 0.2 * x
    df_dy = -np.sin(x) * np.sin(y) + 0.2 * y
    return df_dx, df_dy

x_vals = np.linspace(-5, 5, 100)
y_vals = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x_vals, y_vals)
Z = f(X, Y)

fig = go.Figure(data=[go.Surface(z=Z, x=X, y=Y, colorscale="magma", opacity = 0.7)])
fig.update_layout(
    title="3D Surface Plot of f(x, y)",
    scene=dict(
        xaxis_title="x",
        yaxis_title="y",
        zaxis_title="f(x, y)",
    ),
    autosize=True,
    margin=dict(l=0, r=0, b=0, t=30),
)

def gradient_descent_steps(x0, y0, lr = 0.3) -> tuple[list[float], list[float], list[float]]:
    xs = [x0]
    ys = [y0]
    zs = [f(x0, y0)]
    for i in range(1000):
        x = xs[-1]
        y = ys[-1]
        dx, dy = grad_f(x, y)
        new_x = x - lr * dx
        new_y = y - lr * dy
        xs.append(new_x)
        ys.append(new_y)
        zs.append(f(new_x, new_y))
        if dx**2 + dy**2 < 5e-2:
            break
    return xs, ys, zs

xs, ys, zs = gradient_descent_steps(4.0, 2.5)
fig.add_trace(
    go.Scatter3d(
        x=xs,
        y=ys,
        z=zs,
        mode="markers+lines",
        marker=dict(
            size=3,
            color="cyan",
        ),
        name="Current point",
    )
)

xs, ys, zs = gradient_descent_steps(-1.5, -3.5, lr = 0.5)
fig.add_trace(
    go.Scatter3d(
        x=xs,
        y=ys,
        z=zs,
        mode="markers+lines",
        marker=dict(
            size=3,
            color="orange",
        ),
        name="Current point",
    )
)

print(f"{len(xs)=}")
fig.show()
```
